# Create Heavy Consumption Timeseries Annual Gold Table

## Purpose
Data promotion notebook that creates the gold-layer annual heavy crude consumption time series table from silver-layer cleaned EIA residual fuel consumption data. This notebook prepares consumption data for downstream forecasting and analysis, using residual fuel oil consumption as a proxy for heavy crude consumption.

## Data Flow

**Data Source (Read):**
* `workspace.silver.eia_residual_fuel_consumption_annual` - Cleaned annual residual fuel oil consumption data (Country, Year, productName, unitName, value)

**Data Destination (Write):**
* `workspace.gold.heavy_consumption_timeseries_annual` - Annual heavy crude consumption time series (Country, Year, value)

## Workflow
1. **Extract** - Query silver table for all annual consumption records
2. **Transform** - Select essential columns (Country, Year, value) for time series format
3. **Load** - Write to gold schema with full refresh (overwrite mode)
4. **Validate** - Display sample data and row count verification

## Key Features
* **Full refresh** - Table uses overwrite mode with schema overwrite enabled
* **Schema simplification** - Reduces silver schema to only essential columns for gold layer
* **Time series ready** - Output is structured for direct use in forecasting models and analysis
* **Proxy methodology** - Uses residual fuel oil as a proxy indicator for heavy crude consumption patterns
* **Validation included** - Row count and data preview for quality assurance

## Data Extraction and Gold Table Creation

```python
df = spark.sql("SELECT * FROM workspace.silver.eia_residual_fuel_consumption_annual")

#print the total number of rows
print(df.count())

df.show(10)
```

```python
df = spark.sql("SELECT Country, Year, value FROM workspace.silver.eia_residual_fuel_consumption_annual")

# Write as a Delta table in Unity Catalog (UC)
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.gold.heavy_consumption_timeseries_annual")
```
